```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
import os
import pandas as pd

print(os.getcwd())

PATH = os.getcwd()

print(os.listdir(PATH))

import matplotlib as mpl

plt.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "mathtext.fontset": "cm",
    "font.serif": ["cmr10", "DejaVu Serif", "serif"],
    "axes.formatter.use_mathtext": True,
    "font.size": 12
})
```

# Potência,Tensão,Vazão

```python
file = PATH + "/dadosBancada.txt"

dt = pd.read_csv(file, sep=';', index_col=False)

print(dt.head())

bomba = "bomba(%)"
bomba = dt[bomba]
tensao = "tensao(V)"
tensao = dt[tensao]
vazao = "vazao(l/s)"
vazao = dt[vazao]

plt.plot(bomba, tensao)

plt.grid()
plt.title("Bomba (\%) x Tensao(V)")
plt.xlabel("%")
plt.ylabel("V")
plt.xticks(np.arange(0,100, step=5))
plt.yticks(np.arange(0,11, step=1))

plt.show()
```

```python
plt.plot(bomba, vazao)

plt.grid()
plt.title(r"Bomba (\%) x Vazão($L/s$)")
plt.xlabel("%")
plt.ylabel("V")
plt.xticks(np.arange(0,100, step=5))
plt.yticks(np.arange(0,2.6, step=0.25))

plt.show()
```

# Pulsos, Potêcia

```python
file = PATH + "/dados.csv"
    
MinSamples = 752
MaxSamples = 3000

df = pd.read_csv(
    file,
    header=None,
    names=[
        "pulso_us",
        "pulso_ms",
        "duty",
        "duty_percent"
    ],
    sep='\t'
)

print(df.head())

pulsosus = "pulso_us"
pulsosus = df[pulsosus][MinSamples:MaxSamples]
pulsosms = "pulso_ms"
pulsosms = df[pulsosms][MinSamples:MaxSamples]
duty = "duty"
duty = df[duty][MinSamples:MaxSamples]
dutypercent = "duty_percent"
dutypercent = df[dutypercent][MinSamples:MaxSamples]

plt.plot(dutypercent, pulsosms)
plt.xticks(np.arange(0,110,10))
plt.yticks(np.arange(0,14,1))
plt.grid()

plt.show()
```

```python
# Limita a quantidade de amostras
df = df.iloc[MinSamples:MaxSamples]

# Calcula a média do pulso para cada PWM (%)
media = (
    df.groupby("duty_percent", as_index=False)["pulso_ms"]
      .mean()
      .sort_values("duty_percent")
)

# Gráfico
plt.figure(figsize=(6,4))

plt.plot(
    media["duty_percent"],
    media["pulso_ms"],
    marker='o',
    linewidth=2
)

plt.title("Média dos pulsos em $ms$")
plt.xlabel("PWM ($\%$)")
plt.ylabel("Pulso médio (ms)")

plt.xticks(np.arange(25, 110, 10))
plt.yticks(np.arange(0, 6, 1))

plt.grid(True)

plt.show()
```
