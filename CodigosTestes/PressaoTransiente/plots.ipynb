```python
file = "teste_transiente_06_280726.csv"
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Leitura do arquivo
df = pd.read_csv(file)

# Tempo em segundos
df["tempo_s"] = df["tempo_us"] / 1e6

# Cria figura
fig, ax1 = plt.subplots(figsize=(12, 5))

# Eixo esquerdo - Pressão
ax1.plot(df["tempo_s"], df["pressao"],
         label="Pressão", linewidth=2)

ax1.plot(df["tempo_s"], df["pressao_total"],
         label="Pressão Total", linewidth=2)

ax1.set_xlabel("Tempo (s)")
ax1.set_ylabel("Pressão (kPa)")
ax1.grid(True)

# Eixo direito - Duty Cycle
ax2 = ax1.twinx()

ax2.plot(df["tempo_s"], df["sinal_ad"],
         '--', linewidth=2,
         label="Duty Cycle")

ax2.set_ylabel("Duty Cycle")

# Junta as legendas dos dois eixos
linhas1, labels1 = ax1.get_legend_handles_labels()
linhas2, labels2 = ax2.get_legend_handles_labels()

ax1.legend(linhas1 + linhas2,
           labels1 + labels2,
           loc="best")

plt.title("Resposta da Pressão e Duty Cycle")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Leitura do arquivo CSV
df = pd.read_csv(file)

# Converte o tempo para segundos
df["tempo_s"] = df["tempo_us"] / 1e6

# Cria o gráfico
plt.figure(figsize=(12, 5))

plt.plot(
    df["tempo_s"],
    df["pressao"],
    linewidth=2,
    label="Pressão (P0)"
)

plt.title("Pressão P0")
plt.xlabel("Tempo (s)")
plt.ylabel("Pressão (kPa)")
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
